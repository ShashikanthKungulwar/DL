```python
import numpy as np

inputs = np.array([
      [73, 67,43],
      [91, 88, 64],
      [87, 134, 58],
      [102, 43, 37],
      [69, 96, 70]
],dtype=np.float32)


targets = np.array(
    [
      [56, 70],
      [81, 101],
      [119, 113],
      [22, 37],
      [103, 119]
    ],
    dtype=np.float32
)
```

```python
import torch

inputs = torch.tensor(inputs)
targets = torch.tensor(targets)

print(inputs)
print(targets)
```

```python
weights = torch.randn(2,3,requires_grad = True)
bias = torch.randn(2,requires_grad=True)

print(weights.shape)
print(bias.shape)
```

```python
def model(x):
  return x@weights.T + bias
```

```python
def MSE(y_pred,y_true):
  diff = y_pred - y_true
  return torch.sum(diff*diff)/diff.numel()
```

```python
for i in range(100):
  pred = model(inputs)
  loss = MSE(pred,targets)
  loss.backward()
  print(loss)
  with torch.no_grad():
    lr = 1e-5
    weights -= weights.grad * lr
    bias -= bias.grad * lr
    weights.grad.zero_()
    bias.grad.zero_()
```
